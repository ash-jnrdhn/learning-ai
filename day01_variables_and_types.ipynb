```python
rack_id = "N1R2"
gpu_count = 15
power_draw_kw = 10.4
is_online = True
```

```python
print(rack_id, type(rack_id))
print(gpu_count, type(gpu_count))
print(power_draw_kw, type(power_draw_kw))
print(is_online, type(is_online))
```

```python
print(f"Rack {rack_id} consists of {gpu_count} GPUs, {power_draw_kw} kW, online={is_online}")
```

```python
print(f"{rack_id} has WEKA03 failure as it drew {power_draw_kw+1.4:.5f} KW instead of it's capacity {power_draw_kw:.3f}, hence only {gpu_count-3:.2f} GPUs are working instead of the assigned {gpu_count}")
```
